```python


top_countries = country_funding.head(5)

plt.figure(figsize=(7, 5))
top_countries.plot(kind="bar", color=['darkblue', 'blue','blue','blue','blue'])

plt.title("Top 5 Countries by Total Funding")
plt.xlabel("Country")
plt.ylabel("Total Funding in Billions (USD)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show();
```

```python
usa = companies_df[companies_df["country_code"] == "USA"]

category_counts = usa["category_code"].value_counts().head(10)


category_counts.index = category_counts.index.str.replace("_", " ", regex=False)

category_counts.plot(
    kind="bar",
    title="Top 10 Categories of Startups in USA",
    color=['green','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen','lightgreen']
)

plt.ylabel("Number of Startups")
plt.xticks(rotation=45)
plt.show()
```

```python
usa = companies_df[companies_df["country_code"] == "USA"]

city_roi = usa.groupby("city")["ROI"].sum().sort_values(ascending=False)

city_roi.head(5).plot(
    kind="bar",
    title="Top 5 Cities in USA by ROI"
)

plt.ylabel("Average ROI")
plt.xlabel("City")
plt.xticks(rotation=45)
plt.show()
```
